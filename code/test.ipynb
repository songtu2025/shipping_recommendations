```python
import pandas as pd
DEFAULT_DATE = '20260817'

result_df = pd.read_excel(f'D:/shipping_recommendations2.0/src_data/在途货件/JP在途-预计到仓结果{DEFAULT_DATE}.xlsx')

# 读取 在途货件表
in_transit_shipments = pd.read_excel(f'D:/shipping_recommendations2.0/src_data/在途货件/在途货件{DEFAULT_DATE}.xlsx', sheet_name='在途货件-正常')

# 将merge_df内容放入在途货件表中，这两个表列名完全一致
df = pd.concat([in_transit_shipments, result_df], ignore_index=True)

# 将 在途货件{DEFAULT_DATE}.xlsx的子表'在途货件-正常'内容替换为 df
with pd.ExcelWriter(f'D:/shipping_recommendations2.0/src_data/在途货件/在途货件{DEFAULT_DATE}.xlsx', engine='openpyxl', mode='a', if_sheet_exists='replace') as writer:
    df.to_excel(writer, sheet_name='在途货件-正常', index=False)
```
